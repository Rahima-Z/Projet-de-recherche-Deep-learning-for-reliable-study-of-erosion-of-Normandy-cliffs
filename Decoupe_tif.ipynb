## Import et Variables

```python
from PIL import Image
import tifffile
import matplotlib.pyplot as plt
import numpy as np
import os

filename = "decoupe.tif"
file="decoupe_raster_stride"
data_root =r'C:\Users\Rhm\Documents\M2 SIC\Projet de recherche\dataset\test 3\dataset_stride\Masks_pred\rapport\code\dataset'
os.chdir(data_root)
img= tifffile.imread(filename,key=0)
image_dir = os.path.join(data_root, 'Images')  # dossier contenant les tuiles images (RGB)
mask_dir  = os.path.join(data_root, 'Masks')   # dossier contenant les masques binaires associés
xsize = 256
ysize = 256
stride=64
```

## Découpe : Image*

```python
os.chdir(image_dir)
for x in range(0,int(img.shape[0]),stride):
    for y in range(0,int(img.shape[1]),stride):
        #print("a")
        
        #imgpiece = img.crop((x, y, x+xsize, y+ysize))
        imgpiece=img[x:(x+xsize),y:(y+ysize),:]
        if (np.shape(imgpiece)[0]==256 and np.shape(imgpiece)[1]==256 ) :
            #print(file + "_" + str(x) + "_" + str(y)+"_stride.tif")
            #print(str(x)+","+str(y))
            #print(imgpiece.shape)
            if os.path.exists(file + "_" + str(x) + "_" + str(y)+"_stride.tif"):
                os.remove(file + "_" + str(x) + "_" + str(y)+"_stride.tif")
            
            tifffile.imwrite(file + "_" + str(x) + "_" + str(y)+"_stride.tif",imgpiece) #sauve sous le nom toto_0_0.png, toto_0_1.png, etc.[/code]
```

## Découpe : Masque

```python
filename="mask.tif"
file="mask"
os.chdir(data_root)
img= tifffile.imread(filename,key=0)
os.chdir(mask_dir)
for x in range(0,int(img.shape[0]),stride):
    for y in range(0,int(img.shape[1]),stride):

        imgpiece=img[x:(x+xsize),y:(y+ysize)]
        if (np.shape(imgpiece)[0]==256 and np.shape(imgpiece)[1]==256 ) :
            #print(file + "_" + str(x) + "_" + str(y)+"_stride.tif")
            #print(str(x)+","+str(y))
            #print(imgpiece.shape)
            if os.path.exists(file + "_" + str(x) + "_" + str(y)+"_stride.tif"):
                os.remove(file + "_" + str(x) + "_" + str(y)+"_stride.tif")
            
            tifffile.imwrite(file + "_" + str(x) + "_" + str(y)+"_stride.tif",imgpiece) 
```
